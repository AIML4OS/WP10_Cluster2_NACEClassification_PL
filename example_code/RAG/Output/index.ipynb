```python
prawidlowe = 0 
nieprawidlowe = 0
pierwsze =0
drugie = 0
trzecie =0
for i, row in enumerate(df.iterrows()):
    test =df[["predict_PKD","kod_pkd"]]
    places = str(test['predict_PKD'][i]).split('|')
    if str(places[0]).find(str(test['kod_pkd'][i])) > -1:
        pierwsze +=1
    elif str(places[1]).find(str(test['kod_pkd'][i])) > -1:
        drugie +=1
    elif str(places[2]).find(str(test['kod_pkd'][i])) > -1:
        trzecie +=1
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

RAG_FILE = "../../Dane/ALL_DATA_20.04.2026.csv" # ./source/pkd.csv | ../Dane/ALL_DATA_20.04.2026.csv 


data = pd.read_csv(RAG_FILE,sep=';')
data = data[['PKD_2007','description_PL']]
data = data.rename(columns={
        'PKD_2007': 'kod_pkd', 
        'description_PL': 'opis'
    })
data['kod_pkd'] = data['kod_pkd'].astype(str).str.strip()
data['opis'] = (
        data['opis']
            .str.lower()
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )
data = data[data['opis'].str.len() > 20]
data = data[['kod_pkd', 'opis']].dropna()
data = data.drop_duplicates()

df_train, df_test = train_test_split(data,test_size=0.2,stratify=data["kod_pkd"])
```
